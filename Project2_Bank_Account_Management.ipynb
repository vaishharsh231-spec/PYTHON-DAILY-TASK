# Real-Time Project 2: Bank Account Management

This project is very useful for understanding **`self`, constructors, methods and object-specific data**.

**Requirements**
1. Set the account holder's name and opening balance
2. Deposit money into the account
3. Withdraw money from the account
4. Check the current account balance
5. Maintain each account's balance independently

## Step 1: Define the `BankAccount` class

```python
class BankAccount:
    def __init__(self, holder_name, balance=0):
        self.holder_name = holder_name      # belongs to THIS object
        self.balance = balance              # belongs to THIS object
        print(f"Account created for {self.holder_name} with balance Rs. {self.balance}")

    def deposit(self, amount):
        if amount <= 0:
            print("Invalid amount. Deposit must be greater than 0.")
            return
        self.balance += amount
        print(f"Rs. {amount} deposited. New balance: Rs. {self.balance}")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid amount. Withdrawal must be greater than 0.")
        elif amount > self.balance:
            print(f"Insufficient balance! Available: Rs. {self.balance}")
        else:
            self.balance -= amount
            print(f"Rs. {amount} withdrawn. New balance: Rs. {self.balance}")

    def check_balance(self):
        print(f"{self.holder_name}'s current balance: Rs. {self.balance}")
        return self.balance
```

## Step 2: Create two independent accounts

```python
acc1 = BankAccount("Prem Verma", 5000)
acc2 = BankAccount("Rahul Gupta", 10000)
```

## Step 3: Perform transactions

```python
acc1.deposit(2000)
acc1.withdraw(1500)
acc1.withdraw(10000)     # more than balance
print()
acc2.deposit(500)
acc2.withdraw(3000)
```

## Step 4: Each account keeps its own balance

Changes on `acc1` never affect `acc2`. That is the role of `self`.

```python
acc1.check_balance()
acc2.check_balance()
print("Same class, different objects:", acc1 is acc2)
```

## Step 5: Menu-driven version (optional)

```python
def bank_menu(account):
    while True:
        print("\n1. Deposit  2. Withdraw  3. Check Balance  4. Exit")
        choice = input("Choose an option: ")
        if choice == "1":
            account.deposit(float(input("Amount: ")))
        elif choice == "2":
            account.withdraw(float(input("Amount: ")))
        elif choice == "3":
            account.check_balance()
        elif choice == "4":
            print("Thank you!")
            break
        else:
            print("Invalid choice.")

# Uncomment to try:
# bank_menu(acc1)
```
