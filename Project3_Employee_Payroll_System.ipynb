# Real-Time Project 3: Employee Payroll System

The constructor initializes employee details and methods calculate the salary.

**Requirements**
- Store the employee ID, name and basic salary
- HRA = 20% of the basic salary
- DA = 10% of the basic salary
- PF deduction = 12% of the basic salary
- Net salary = Basic Salary + HRA + DA - PF

> Note: These percentages are used for demonstration purposes. Actual payroll rules depend on the organization and applicable regulations.

## Step 1: Define the `Employee` class

```python
class Employee:
    HRA_RATE = 0.20
    DA_RATE = 0.10
    PF_RATE = 0.12

    def __init__(self, emp_id, name, basic_salary):
        self.emp_id = emp_id
        self.name = name
        self.basic_salary = basic_salary

    def calculate_hra(self):
        return self.basic_salary * self.HRA_RATE

    def calculate_da(self):
        return self.basic_salary * self.DA_RATE

    def calculate_pf(self):
        return self.basic_salary * self.PF_RATE

    def calculate_net_salary(self):
        return (self.basic_salary
                + self.calculate_hra()
                + self.calculate_da()
                - self.calculate_pf())

    def display_payslip(self):
        print("=" * 40)
        print("            PAYSLIP")
        print("=" * 40)
        print(f"Employee ID   : {self.emp_id}")
        print(f"Name          : {self.name}")
        print("-" * 40)
        print(f"Basic Salary  : Rs. {self.basic_salary:,.2f}")
        print(f"HRA (20%)     : Rs. {self.calculate_hra():,.2f}")
        print(f"DA (10%)      : Rs. {self.calculate_da():,.2f}")
        print(f"PF (12%)      : - Rs. {self.calculate_pf():,.2f}")
        print("-" * 40)
        print(f"Net Salary    : Rs. {self.calculate_net_salary():,.2f}")
        print("=" * 40)
```

## Step 2: Create employee objects

```python
emp1 = Employee(1001, "Prem Verma", 30000)
emp2 = Employee(1002, "Sneha Rani", 45000)
emp3 = Employee(1003, "Karan Mehta", 60000)
```

## Step 3: Display payslips

```python
emp1.display_payslip()
emp2.display_payslip()
```

## Step 4: Payroll summary for all employees

```python
employees = [emp1, emp2, emp3]

print(f"{'ID':<7}{'Name':<14}{'Basic':>10}{'HRA':>10}{'DA':>10}{'PF':>10}{'Net':>12}")
for e in employees:
    print(f"{e.emp_id:<7}{e.name:<14}{e.basic_salary:>10,.0f}{e.calculate_hra():>10,.0f}"
          f"{e.calculate_da():>10,.0f}{e.calculate_pf():>10,.0f}{e.calculate_net_salary():>12,.0f}")

total = sum(e.calculate_net_salary() for e in employees)
print(f"\nTotal payroll (net): Rs. {total:,.2f}")
```
