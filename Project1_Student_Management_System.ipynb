# Real-Time Project 1: Student Management System

We combine **class, object, constructor and methods** to build a practical mini-project.

**Features**
- Create student objects
- Initialize student details using a constructor
- Store student marks
- Calculate total marks and average marks
- Calculate the student's grade
- Display the complete student report

## Step 1: Define the `Student` class

```python
class Student:
    # Constructor: runs automatically when an object is created
    def __init__(self, roll_no, name, marks):
        self.roll_no = roll_no
        self.name = name
        self.marks = marks          # dictionary {subject: marks}

    # Method 1: total marks
    def calculate_total(self):
        return sum(self.marks.values())

    # Method 2: average marks
    def calculate_average(self):
        return self.calculate_total() / len(self.marks)

    # Method 3: grade based on average
    def calculate_grade(self):
        avg = self.calculate_average()
        if avg >= 90:
            return "A+"
        elif avg >= 80:
            return "A"
        elif avg >= 70:
            return "B"
        elif avg >= 60:
            return "C"
        elif avg >= 50:
            return "D"
        else:
            return "Fail"

    # Method 4: complete report
    def display_report(self):
        print("=" * 40)
        print("          STUDENT REPORT")
        print("=" * 40)
        print(f"Roll No : {self.roll_no}")
        print(f"Name    : {self.name}")
        print("-" * 40)
        for subject, mark in self.marks.items():
            print(f"{subject:<15}: {mark}")
        print("-" * 40)
        print(f"Total   : {self.calculate_total()}")
        print(f"Average : {self.calculate_average():.2f}")
        print(f"Grade   : {self.calculate_grade()}")
        print("=" * 40)
```

## Step 2: Create student objects

```python
student1 = Student(101, "Prem Verma", {"Python": 92, "SQL": 88, "Power BI": 85, "Maths": 90})
student2 = Student(102, "Amit Sharma", {"Python": 65, "SQL": 58, "Power BI": 70, "Maths": 62})
student3 = Student(103, "Neha Singh", {"Python": 45, "SQL": 40, "Power BI": 52, "Maths": 38})
```

## Step 3: Display the reports

```python
student1.display_report()
student2.display_report()
student3.display_report()
```

## Step 4: Handle many students with a list of objects

```python
students = [student1, student2, student3]

print(f"{'Roll':<6}{'Name':<15}{'Total':<8}{'Average':<10}{'Grade'}")
for s in students:
    print(f"{s.roll_no:<6}{s.name:<15}{s.calculate_total():<8}{s.calculate_average():<10.2f}{s.calculate_grade()}")
```

## Step 5: Take input from the user (optional)

```python
def create_student_from_input():
    roll_no = int(input("Enter roll number: "))
    name = input("Enter name: ")
    n = int(input("How many subjects? "))
    marks = {}
    for _ in range(n):
        subject = input("  Subject name: ")
        marks[subject] = float(input(f"  Marks in {subject}: "))
    return Student(roll_no, name, marks)

# Uncomment to try:
# new_student = create_student_from_input()
# new_student.display_report()
```
